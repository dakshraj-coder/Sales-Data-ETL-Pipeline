# Data exploration
A first look at the raw CSV files before building the pipeline.
Run this notebook from inside the `notebooks/` folder.

```python
import pandas as pd
from pathlib import Path

raw_dir = Path('../data/raw')
frames = [pd.read_csv(f, dtype=str) for f in sorted(raw_dir.glob('*.csv'))]
for f, df in zip(sorted(raw_dir.glob('*.csv')), frames):
    print(f.name, df.shape)
```

Column names differ between files (`order_id` vs `Order Id`). `extract.py` tidies them.

```python
for df in frames:
    df.columns = [c.strip().lower().replace(' ', '_') for c in df.columns]
raw = pd.concat(frames, ignore_index=True)
raw.head()
```
